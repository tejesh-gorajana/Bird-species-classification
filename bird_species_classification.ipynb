# Bird Species Classification using EfficientNet-B2 (pretrained)

This notebook uses the pretrained Hugging Face model **`dennisjooo/Birds-Classifier-EfficientNetB2`**, an EfficientNet-B2 model fine-tuned (transfer learning) for bird species classification.

The model is **not trained here**. It is downloaded automatically on the first run (internet needed) and cached on your computer, so no model file is stored in the project folder.

Workflow: load model → inspect model → predict a bird image → show top-5 predictions.

## 1. Import required libraries

```python
import os

import matplotlib.pyplot as plt
from PIL import Image, ImageOps
from transformers import pipeline
```

## 2. Define configuration

```python
MODEL_ID = "dennisjooo/Birds-Classifier-EfficientNetB2"

# Put a bird photo in this same folder and write its file name here
IMAGE_PATH = "my_bird.jpg"
```

## 3. Load the pretrained model
The first run downloads the model, so it can take a minute.

```python
classifier = pipeline("image-classification", model=MODEL_ID)
print("Model loaded:", MODEL_ID)
```

## 4. Inspect the model

```python
labels = classifier.model.config.id2label
print("Number of bird species:", len(labels))
print("Parameters:", f"{classifier.model.num_parameters():,}")
print("First 10 species:", [labels[i].title() for i in range(10)])
```

## 5. Load a bird image
This cell loads your photo. If the file is missing it tells you instead of crashing.

```python
if os.path.exists(IMAGE_PATH):
    image = ImageOps.exif_transpose(Image.open(IMAGE_PATH)).convert("RGB")
    plt.imshow(image)
    plt.axis("off")
    plt.title("Input image")
    plt.show()
else:
    image = None
    print(f"'{IMAGE_PATH}' not found. Copy a bird photo into this folder and set IMAGE_PATH above.")
```

## 6. Predict the bird species

```python
if image is not None:
    results = classifier(image, top_k=5)

    best = results[0]
    print(f"Predicted species: {best['label'].title()}")
    print(f"Confidence: {best['score'] * 100:.2f}%")
else:
    print("Skipped: no image loaded.")
```

## 7. Show the top 5 predictions

```python
if image is not None:
    names = [r["label"].title() for r in results][::-1]
    scores = [r["score"] * 100 for r in results][::-1]

    plt.figure(figsize=(8, 4))
    plt.barh(names, scores)
    plt.xlabel("Confidence (%)")
    plt.title("Top 5 predictions")
    plt.tight_layout()
    plt.show()
else:
    print("Skipped: no image loaded.")
```
